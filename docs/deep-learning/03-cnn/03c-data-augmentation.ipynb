```python
import matplotlib.pyplot as plt

def savefig(filename: str, caption_topic="", caption=""):
    plt.savefig(f"./plots/{filename}.svg", bbox_inches="tight")
    plt.close("all")

    if caption_topic:
        fig_caption = f"**{caption_topic}.** {caption}"
    else:
        fig_caption = caption

    print(f""":::{{figure}} ./plots/{filename}.svg
---
name: {filename}
width: 100%
align: center
---
{fig_caption}
:::""")
```

```python
plt.plot([1, 2, 3])
savefig("hello", caption_topic="hello", caption="world")
```

:::{figure} ./plots/hello.svg
---
name: hello
width: 100%
align: center
---
**hello.** world
:::
