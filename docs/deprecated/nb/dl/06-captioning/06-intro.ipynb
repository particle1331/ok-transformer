# Attention & Image Captioning

Our objective is to train a model $f$ that takes an image $\boldsymbol{\mathsf{x}} \in \mathbb{R}^d$ and outputs a caption $f(\boldsymbol{\mathsf{x}}) \in \mathbb{N}^{T_{\text{max}}}$ for some ${T_{\text{max}}} \in \mathbb{N}.$ Hence, we have a mapping from images to text sequences. To do this, we implement the paper [*Show, Attend, and Tell*](https://arxiv.org/pdf/1502.03044) {cite}`showattendtell`. Our presentation is based on this [tutorial](https://github.com/sgrvinod/a-PyTorch-Tutorial-to-Image-Captioning). The general setup is shown in {numref}`06-img-captioning`.

This model learns where to look via the **attention mechanism**. As the caption is generated word by word, the model's gaze shifts across the image. The attention mechanism allows the model to focus on the part of the image most relevant to the word it is going to utter next.
